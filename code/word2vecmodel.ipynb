```python
from pathlib import Path
from urllib.request import urlretrieve

from gensim.models import KeyedVectors
```

## Download the Pretrained Google News Model

The original Google News Word2Vec model is large (about **1.5 GB compressed**),
so the download may take some time.

The file is downloaded only if it is not already present.


```python
MODEL_URL = "https://s3.amazonaws.com/dl4j-distribution/GoogleNews-vectors-negative300.bin.gz"
MODEL_PATH = Path("GoogleNews-vectors-negative300.bin.gz")

if not MODEL_PATH.exists():
    print("Downloading Google News Word2Vec vectors...")
    urlretrieve(MODEL_URL, MODEL_PATH)
    print("Download complete.")
else:
    print("Model file already exists. Skipping download.")
```

## Load Word Vectors

`limit=500_000` loads only the first 500,000 vectors to reduce memory usage.

Each word is represented by a **300-dimensional dense vector**.


```python
model = KeyedVectors.load_word2vec_format(
    MODEL_PATH,
    binary=True,
    limit=500_000
)

print(f"Loaded vocabulary size: {len(model.key_to_index):,}")
print(f"Embedding dimension: {model.vector_size}")
```

## Inspect a Word Vector

Instead of printing all 300 values, display the vector shape and the first
10 dimensions.


```python
cricket_vector = model["cricket"]

print("Vector shape:", cricket_vector.shape)
print("First 10 dimensions:")
print(cricket_vector[:10])
```

## Find Similar Words

`most_similar()` uses cosine similarity to find words whose vectors are close
to the query word.


```python
model.most_similar("man", topn=10)
```

```python
model.most_similar("cricket", topn=10)
```

```python
model.most_similar("facebook", topn=10)
```

## Cosine Similarity Between Two Words

Cosine similarity is commonly used to compare Word2Vec vectors.

- Closer to `1` → more similar direction
- Around `0` → weak directional similarity
- Negative → vectors point in more opposite directions


```python
model.similarity("man", "woman")
```

```python
model.similarity("man", "PHP")
```

## Find the Odd Word Out

`doesnt_match()` returns the word whose vector is least consistent with the
other words in the group.


```python
model.doesnt_match(["PHP", "java", "monkey"])
```

## Word-Vector Analogy

A classic Word2Vec demonstration uses vector arithmetic such as:

\[
\text{king} - \text{man} + \text{woman} \approx \text{queen}
\]

Using word names directly lets Gensim exclude the input words from the returned
nearest neighbors.


```python
model.most_similar(
    positive=["king", "woman"],
    negative=["man"],
    topn=10
)
```

## Another Analogy Example

The same vector-arithmetic idea can be explored with country/currency-related
terms:

\[
\text{INR} - \text{India} + \text{England}
\]

The result is an exploratory property of the learned embedding space, not a
guaranteed symbolic rule.


```python
model.most_similar(
    positive=["INR", "England"],
    negative=["India"],
    topn=10
)
```

## Key Takeaway

Word2Vec represents words as dense numerical vectors. Because the vectors are
learned from word co-occurrence patterns, semantic and syntactic relationships
can emerge as geometric relationships in the embedding space.

For the full training mathematics, refer to the accompanying workshop resources
covering **CBOW, Skip-gram, negative sampling, sigmoid, loss, and weight updates**.
